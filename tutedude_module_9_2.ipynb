```python
plt.figure(figsize=(10, 10))          # create figure FIRST
plt.xlabel('Months')
plt.ylabel('Total sales')
plt.bar(np.arange(m.shape[0]) - 0.2, m.values, color='green', label='Mint Chip Choco', width=0.2)
plt.bar(np.arange(d.shape[0]),       d.values, color='blue',  label='85% Dark Bars',      width=0.2)
plt.bar(np.arange(p.shape[0]) + 0.2, p.values, color='red',   label='Peanut Butter Cubes', width=0.2)

# Put ticks at the MIDDLE of each group (not -0.2)
positions = np.arange(len(x))
plt.xticks(positions, x, rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#stacked bar chart-->keep the bar in stack manner,one bar upon other
```

```python
# Stacked Bar Chart
# Instead of Keeping the Bars in Side by Side Direction , we can keep the Bars in the Stacked Manner i.e, One on Top of Other and in order to Figure out which category is having Maximum Sale , we can have a look over the Length of the Bar
# Stacked Bar Chart---> bar()---> parameter---> bottom ---> This bottom parameter specific that what bar i want to keep below my current bar
# Red Bar
# Green Bar---> bottom---> Red Bar
# Blue Bar ---> bottom---> Green Bar + Red Bar
```

```python
plt.figure(figsize=(10, 10))          # create figure FIRST
plt.xlabel('Months')
plt.ylabel('Total sales')
plt.bar(m.index, m.values, color='green', label='Mint Chip Choco', width=0.2)
plt.bar(d.index,       d.values, color='blue',  label='85% Dark Bars',      width=0.2,bottom=m.values)
plt.bar(p.index, p.values, color='red',   label='Peanut Butter Cubes', width=0.2,bottom=m.values+d.values)

# Put ticks at the MIDDLE of each group (not -0.2)
positions = np.arange(len(x))
# plt.xticks(positions, x, rotation=45)
plt.legend()
plt.tight_layout()
plt.show()
```
