```python
from datetime import date
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def initialize_person(cls,name,year):
        return cls(name, date.today().year - year)

    @staticmethod
    def check_age_eligibility(age):
        if age >= 18:
            return 'Eligible' 

    def show_person(self):
        print(self.name, self.age)

person1 = Person.initialize_person('Chintamani', 1999)
person1.show_person()
```

```python
# ## Inheritance


# #### Inheritance is a Fundamental concept in Object Oriented Programming because of which the 
# child class is able to inherit the properties OR Attributes and functionalities of the parent 
# class


# #### Just like a  child is going to inherit the properties from parent and parent is going to 
# inherit the properties from the grandparent and definitieily in this hierarchy the grandchild can 
# have some features different from the parent and grandparent as well .

# #### In Python typically we get to see 3 Types of Inheritance :
# 1. Single Inheritance
# 2. Multilevel Inheritance
# 3. Multiple Inheritance

#### 1) Single Inheritance --> The Features or the Attributes from one parent class will be getting inherited into one child class .

### Parent Class (Properties--->Attributes and Functionalities)
### |
### |
### Child Class

# ## 2)Multilevel Inheritance--> The properties or Methods and Attributes of the grandparent class 
# can be inherited by the parent class and then the properties of the parent class will be 
# inherited by the child class . In a proper heirarchy the Inheritance process is going to work for 
# us .


# ### Grandparent (Properties)
# ###  |
# ###  |
# ### Parent (Properties)
# ###  |
# ###  |
# ### Child |

### 3)Multiple Inheritance --> In Multiple Inheritance , The properties of two parent class or multiple parent class will be getting inherited into the child class

### Parent_1 (Properties) + Parent_2 (Properties) + Parent_3 (Properties).....
### |
### |
### Child
```

```python
class Grandparent:
    def __init__(self, name, locker_key):
        self.name = name
        self._locker_key = locker_key

    def locker_key_info(self):
        return self._locker_key

gp1 = Grandparent('Dattatraya', 1234)
print(gp1.locker_key_info())
# print(dir(gp1)) # (Commented out to keep output clean, but you can leave it)

# --- FIX 1: Add (Grandparent) here to establish inheritance ---
class Parent(Grandparent): 
    def __init__(self, name, locker_key):
        # Now super() refers to Grandparent, and this will work correctly
        super().__init__(name, locker_key)

    def __str__(self):
        # self.name and self._locker_key now exist because of the super() call
        return f"Name: {self.name}, locker_key: {str(self._locker_key)}"

    @classmethod
    def return_object(cls, name, locker_key):
        return cls(name, locker_key)

p1 = Parent("Mahesh", 1234)
print(p1)
print(p1._locker_key)
p2 = Parent.return_object('Mahesh',1234)
print(p2)
```

```python
class Bankaccount:
    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount > 0 :
            self.__balance += amount
        else:
            print("Amount should be positive")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount

bacc1 = Bankaccount('Chintamani', 0)
print(bacc1.get_balance())
bacc1.deposit(1000)
print(bacc1.get_balance())
bacc1.withdraw(500)
print(bacc1.get_balance())
```

```python
# How can we achieve abstraction in python

# 1. Abstract class --> A class that cannot be instantiated or that means for an abstract class we 
# cant create the object of this class .

# 2. Abstract method --> Abstract method just because it cant accessed by the object of the abstract class 
# itself thats why we have to call this abstract method in our child class and the implementation of the abstract 
# method is not provided inside the abstract class but must be provided in the child class .
```

```python
from abc import ABC, abstractmethod
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass

    @abstractmethod
    def stop(self):
        pass

class Car(Vehicle):
    def start(self):
        print("The car starts with ignition key")

    def stop(self):
        print("The car starts with pressing break paddle")

class ElectricScooter(Vehicle):
    def start(self):
        print("The ElectricScooter starts with start button")

    def stop(self):
        print("The car starts with pressing stop button")


car = Car()
car.start()

electricScooter = ElectricScooter()
electricScooter.start()

veh = Vehicle()
```

```python
# Magic methods
# Magic methods are also called as Dunder methods bacause they have double underscores before and after their name
# Special methods in python that allow us to customize how objects behave
# These methods help us define:
# 1)Object initialization __init__ 
# 2)String representation __rep__, __str__
# 3)Arithmatic operations __add__, __sub__
# 4)Comparison operators __eq__, __lt__, __gt__
# 5)Length, indexing , iteration (len,getitem,etc)

# why to use magic methods:

# 1)Makes objects behave like built in types (eg, lists , integers)

# 2)Improves Readability by allowing basic operations object1 + object2

# 3)Simplifies code by replacing verbose function calls with natural syntax .
```

```python
class Avenger:
    def __init__(self, name, strength):
        self.name = name
        self.strength = strength

    def __add__(self, other):
        return f"{self.name +' & ' +other.name} Strength: {self.strength + other.strength}"

    def __str__(self):
        return f"Name: {self.name}, Strength: {self.strength}"

avg1 = Avenger('Hulk',2)
print(avg1)
avg2 = Avenger('Ironman',20)
print(avg2)
print(avg1 + avg2)
```

```python
# Operartor overloading allow us to define how operators (*,-,=,+,...etc)

# Operator ---> + ---> Method name--> add(self,other)
# Operator ---> - ---> Method name ---> sub(self,other)
# Operator ---> * ---> Method name ---> mul(self,other)
# Operator ---> / ---> Method name ---> truediv(self,other)
# Operator ---> == --> Method name --> eq(self,other)
# Operator ---> > --> Method name --> gt(self,other)
```

```python
class Avenger:
    def __init__(self, name, power):
        self.name = name
        self.power = power

    def __add__(self, other):
        return Avenger(self.name+ ' & '+other.name, self.power + other.power)

    def __sub__(self, other):
        return Avenger(self.name+ ' & '+other.name, abs(self.power - other.power))

    def __mul__(self, other):
        return Avenger(self.name+ ' & '+other.name, self.power * other.power)

    def __truediv__(self, other):
        return Avenger(self.name+ ' & '+other.name, self.power / other.power)

    def __str__(self):
        return f"Name: {self.name} , Power: {self.power}"

    def __gt__(self, other):
        return  self.power > other.power

    def __lt__(self, other):
        return  self.power < other.power

    def __eq__(self, other):
        return  self.power < other.power

avg1 = Avenger("Ironman", 100)
avg2 = Avenger("Thor", 200)
avg3 = Avenger("Hulk", 200)
print(avg1)
print(avg2)
print(avg1 + avg2)
print(avg1 * avg2)
print(avg1 > avg2)
print(avg1 < avg2)
print(avg1 / avg2)
print(avg1 - avg2)
print(avg2 - avg3)
```
