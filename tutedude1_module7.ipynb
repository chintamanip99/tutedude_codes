```python
# Numpy is a Fundamental Package for scientific computing in python . It is a Python Library that 
# provides a Multi Dimensional Array object and various derived objects such as Masked Arrays or Matrices 
# and an assortment of routines for fast operations on Arrays including Mathematical , Logical , Shape Manipulation , Sorting , 
# Selecting , I/O , Discrete Fourier Transform , Basic Linear Algebra , Stastical Operations , Random Simulation , etc .
# At the core of the Numpy Package is the ndimensional Array object .

## WHY Numpy ?

# ### 2008-2009, Data science ---> Matlab , R Language

# ## Data Types in Python --> List , Dictionary--> Slow
# Syntax--> Python --> Easy
# ## We needed Speed and Less Memory to work with the huge amount of data .

# Numpy --> C language --> wrapped ---> Python language
# ## Numpy--> syntax---> Easy for beginners--> Performing Scientific Calculations ---> Creating Multi
# Dimensional Arrays---> Speed is faster for applying Operations ---> Less Memory

# Libraries like Scikit Learn , Tensorflow and Pytorch require input data in
# Numpy Arrays or Tensors .

# Numpy Arrays Vs Python Sequences

# Numpy Arrays have a fixed size at creation , unlike Python Lists which grow dynamically . Chaning the size of
# an ndim array will create a new array and delete the original .

# The elements in a Numpy Array are required to be of the same data type and thus will be of the same size in memory .

# Numpy Arrays facilitate advanced Mathematical and other types of operations on large number of data . 
# Typically , such operations are executed more effeiciently with less code than is possible using python's built in sequences .

```

```python
#stacking is join two arrays together ,row by row--> Horizontal stacking, col by col--> Vertical stacking
#size of array must be same
```

```python
arr1 = np.arange(1, 13).reshape(3, 4)
arr2 = np.arange(13, 25).reshape(3, 4)
print(arr1)
print(arr2)
horizontal_array = np.hstack((arr1, arr2))
print(horizontal_array)
vertical_array = np.vstack((arr1, arr2))
print(vertical_array)
```

```python
print(horizontal_array)
print("\n")
a,b = np.hsplit(horizontal_array, 2)
print(a)
print(b)
print(100*"*")
print(vertical_array)
print("\n")
a,b = np.hsplit(vertical_array, 2)
print(a)
print(b)
```
