```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('stu2.csv')

cat_cols = ['Gender', 'City', 'Profession', 'Sleep Duration',
            'Dietary Habits', 'Degree']
df[cat_cols] = df[cat_cols].astype('category')
df['CGPA'] = df['CGPA'].astype('float32')
df['Financial Stress'] = df['Financial Stress'].astype('float32')
df.info()

# Average CGPA per City, sorted for a clean horizontal bar chart
data = df.groupby('City', observed=True)['CGPA'].mean().sort_values()

x = data.index.astype(str)   # City names
y = data.values              # mean CGPA

plt.figure(figsize=(8, 10))
plt.xticks(rotation=90, fontsize=8)
plt.bar(x, y, color='steelblue')
plt.xlabel('Average CGPA')
plt.ylabel('City')
plt.title('Average CGPA by City')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 10))
plt.barh(x, y, color='steelblue')
plt.xlabel('Average CGPA')
plt.ylabel('City')
plt.title('Average CGPA by City')
plt.tight_layout()
plt.show()


#Task3
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------- Load & clean ----------
sales = pd.read_csv('Chocolate Sales.csv')

sales['Sales Person'] = sales['Sales Person'].astype('category')
sales['Country']      = sales['Country'].astype('category')
sales['Product']      = sales['Product'].astype('category')

sales['Date'] = pd.to_datetime(sales['Date'])
sales['Date'] = sales['Date'].dt.month_name()

sales['Amount'] = sales['Amount'].str.strip('$').str.replace(',', '', regex=False)
sales['Amount'] = sales['Amount'].astype('int32')
sales['Boxes Shipped'] = sales['Boxes Shipped'].astype('int32')
sales['Date'] = sales['Date'].astype('category')

sales.info()

# ---------- Pivot: Month × Product ----------
pivot = (sales.groupby(['Date', 'Product'], observed=True)['Amount']
              .sum()
              .unstack(fill_value=0))

month_order = ['January','February','March','April','May','June',
               'July','August','September','October','November','December']
present = [m for m in month_order if m in pivot.index]
pivot = pivot.loc[present]

# ---------- 1. Grouped bar chart ----------
ax = pivot.plot(kind='bar', figsize=(12, 6), width=0.8,
                colormap='tab10', edgecolor='black', linewidth=0.4)
ax.set_xlabel('Month')
ax.set_ylabel('Total Sales ($)')
ax.set_title('Monthly Sales by Product')
ax.legend(title='Product', bbox_to_anchor=(1.02, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()







print(pivot)
```
